```python
import math
from phe import paillier

public_key, private_key = paillier.generate_paillier_keypair()

user_lat = 28.52345
user_lon = -80.68309 
centre_lat = 28.52400
centre_lon = -80.68200
r = 100.0 #in km
R = 6371000
scaling_Factor = 1000

def encrypt_user_location(user_lat,user_lon):

    lat_radian = math.radians(user_lat)
    lon_radian = math.radians(user_lon)

    lat_radian_cos = math.cos(lat_radian)
    lat_radian_sin = math.sin(lat_radian)
    lon_radian_cos = math.cos(lon_radian)

    c1 = public_key.encrypt(lat_radian_cos * scaling_Factor)
    c2 = public_key.encrypt(lat_radian_sin * scaling_Factor)
    c3 = public_key.encrypt(lat_radian_cos * lon_radian_cos * scaling_Factor)
    c4 = public_key.encrypt(lat_radian_cos * lat_radian_sin * scaling_Factor)

    return (c1,c2,c3,c4)

def compute(c,centre_lat,centre_lon):
    c1,c2,c3,c4 = c

    centre_lat_radian = math.radians(centre_lat)
    centre_lon_radian = math.radians(centre_lon)

    lat_radian_cos = math.cos(centre_lat_radian * scaling_Factor)
    lat_radian_sin = math.sin(centre_lat_radian * scaling_Factor)
    lon_radian_cos = math.cos(centre_lon_radian * scaling_Factor)

    enc_c = (c1 * round(lat_radian_cos) + 
             c2 * round(lat_radian_sin) + 
             c3 * round(lat_radian_cos * lon_radian_cos) + 
             c4 * round(lat_radian_cos * lat_radian_sin))
    
    #enc_c = (c1 * lat_radian_cos + 
     #        c2 * lat_radian_sin + 
      #       c3 * lat_radian_cos * lon_radian_cos + 
       #      c4 * lat_radian_cos * lat_radian_sin)

    return enc_c

def evaluate(enc_c,R,private_key):

    decrypt_c = private_key.decrypt(enc_c)

    #d = 2 * R * math.asin(math.sqrt((1 - decrypt_c) / 2))
    d = 2 * R * math.asin(math.sqrt(1 - decrypt_c**2))
    return d

def result(d,r):
    if d <= r:
        print("User is inside the geofence.")
    else:
        print("User is outside the geofence.")

# Run the geofencing check
encrypted_location = encrypt_user_location(user_lat, user_lon)
enc_c = compute(encrypted_location, centre_lat, centre_lon)
d = evaluate(enc_c, R, private_key)
result(d, r)
```
